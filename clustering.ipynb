```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.cluster import AgglomerativeClustering
```

```python
iris=load_iris()
df=pd.DataFrame(iris.data,columns=iris.feature_names)
df.head()
x=df
scaler=StandardScaler()
x_scaled=scaler.fit_transform(x)
```

```python
# apply kmeans
kmeans=KMeans(n_clusters=3,random_state=42)
kmeans_labels=kmeans.fit_predict(x_scaled)
df["kmeans_labels"]=kmeans_labels
df.head()
```

```python
plt.figure(figsize=(10,6))
plt.scatter(
    x_scaled[:,0],
    x_scaled[:,1],
    c=kmeans_labels,
    cmap="viridis"
)
plt.xlabel("sepal length")
plt.ylabel("sepal width")
plt.title("K-Means Clustering on iris dataset")

plt.colorbar(label="cluster")
plt.show()
```

```python
# apply herirchical clustring
hc=AgglomerativeClustering(n_clusters=3)
hc_labels=hc.fit_predict(x_scaled)
df["hierarchical_cluster"]=hc_labels
df.head()
```

```python
plt.figure(figsize=(10,6))
plt.scatter(
    x_scaled[:,0],
    x_scaled[:,1],
    c=kmeans_labels,
    cmap="rainbow"
)
plt.xlabel("sepal length")
plt.ylabel("sepal width")
plt.title("heriraechical Clustering on iris dataset")

plt.colorbar(label="cluster")
plt.show()
```

```python
print("KMeans and Hierarchical Clustering were successfully applied to the Iris dataset.")
print("Both algorithms grouped the flowers into 3 clusters based on feature similarity.")
```

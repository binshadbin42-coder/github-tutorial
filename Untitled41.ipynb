```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("puneet6060/intel-image-classification")

print("Path to dataset files:", path)
```

```python
import os

```

```python
os.listdir(path)
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout
from tensorflow.keras.preprocessing.image import ImageDataGenerator
import matplotlib.pyplot as plt
import numpy as np
from tensorflow.keras.preprocessing import image
```

```python
train_dir = os.path.join(path, "seg_train", "seg_train")

classes = sorted(os.listdir(train_dir))

print("Classes:", classes)
print("Number of Classes:", len(classes))
```

```python
for cls in classes:
    folder = os.path.join(train_dir, cls)
    print(f"{cls}: {len(os.listdir(folder))} images")
```

```python
total = sum(len(os.listdir(os.path.join(train_dir, cls))) for cls in classes)

print("Total Images:", total)
```

```python
from PIL import Image

sample = os.path.join(train_dir, classes[0], os.listdir(os.path.join(train_dir, classes[0]))[0])

img = Image.open(sample)

print("Image Size:", img.size)
print("Image Mode:", img.mode)
print("Image Format:", img.format)
```

```python
train_datagen=ImageDataGenerator(rescale=1./255,)

test_datagen=ImageDataGenerator(rescale=1./255)
```

```python
train_dir = os.path.join(path, "seg_train", "seg_train")
test_dir = os.path.join(path, "seg_test", "seg_test")

train_generator=train_datagen.flow_from_directory(train_dir,
                                                 target_size=(150,150),
                                                 batch_size=32,
                                                 class_mode='categorical',
                                                 shuffle=True)

val_generator=test_datagen.flow_from_directory(test_dir,
                                                 target_size=(150,150),
                                                 batch_size=32,
                                                 class_mode='categorical',
                                                shuffle=False)
```

```python
model=Sequential()
#1 lyr
model.add(Conv2D(32,(3,3),activation='relu',input_shape=(150,150,3)))
model.add(MaxPooling2D(2,2))
# 2 lyr
model.add(Conv2D(64,(3,3),activation='relu'))
model.add(MaxPooling2D(2,2))
#3 lyr
model.add(Conv2D(128,(3,3),activation='relu'))
model.add(MaxPooling2D(2,2))
# flatten
model.add(Flatten())
# fully connected lyr
model.add(Dense(512,activation='relu'))
model.add(Dropout(0.5))
# output lyr
model.add(Dense(train_generator.num_classes,activation='softmax'))
```

```python
compile_model=model.compile(loss='categorical_crossentropy',
              optimizer='adam',
              metrics=['accuracy'])
```

train the model

```python
history =  model.fit(train_generator,
                    steps_per_epoch=train_generator.samples // train_generator.batch_size,
                    epochs=20,
                    validation_data=val_generator)
```

testing the model


```python
plt.plot(history.history['accuracy'])
plt.plot(history.history['val_accuracy'])
plt.title("Model Accuracy")
plt.ylabel("Accuracy")
plt.xlabel("Epoch")
plt.legend(['Train','Validation'])
plt.show()
```

```python
test_loss,test_acc =model.evaluate(val_generator)
print("Test Accuracy:",test_acc)
```

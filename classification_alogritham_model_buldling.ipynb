```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score,classification_report
```

```python
data=load_breast_cancer()
x=pd.DataFrame(data.data,columns=data.feature_names)
y=data.target
print("datashape:",x.shape)
print("missing values:|n",x.isnull().sum().sum())
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
scaler=StandardScaler()
x_train=scaler.fit_transform(x_train)
x_test=scaler.transform(x_test)
```

```python
scaler=StandardScaler()
x_train_scaled=scaler.fit_transform(x_train)
x_test_scaled=scaler.transform(x_test)
```

```python
model={
    "LogisticRegression":LogisticRegression(),
    "DecisionTreeClassifier":DecisionTreeClassifier(),
    "RandomForestClassifier":RandomForestClassifier(),
    "SVC":SVC(),
    "KNeighborsClassifier":KNeighborsClassifier()
}
```

```python
results = {}

for name, current_model in model.items():

    # Scaling required for LR, SVM, KNN
    if name in ["LogisticRegression", "SVC", "KNeighborsClassifier"]:
        current_model.fit(x_train_scaled, y_train)
        y_pred = current_model.predict(x_test_scaled)
    else:
        current_model.fit(x_train, y_train)
        y_pred = current_model.predict(x_test)

    accuracy = accuracy_score(y_test, y_pred)
    results[name] = accuracy

    print(f"\n{name}")
    print("-" * 40)
    print("Accuracy:", round(accuracy * 100, 2), "%")
    print(classification_report(y_test, y_pred))
```

```python
comparison=pd.DataFrame(
    results.items(),
    columns=["Model","Accuracy"]
)
comparison = comparison.sort_values(
    by="Accuracy",ascending=False
)
print('\nmodel comparison')
print(comparison)
```

```python
#best and wrost model
best_model=comparison.iloc[0]
wrost_model=comparison.iloc[-1]
print("\nbest model:",best_model)
print("\nwrost model:",wrost_model)
```

```python
# Target variable
y = df['price']

# Feature variables
X = df.drop('price', axis=1)

print("Feature Shape :", X.shape)
print("Target Shape :", y.shape)
```

```python
cat_cols = X.select_dtypes(include='object').columns

print(cat_cols)
```

```python
X = pd.get_dummies(X, drop_first=True)

X.head()
```

```python
print("Shape after Encoding:", X.shape)
```

```python
# Combine X and y
data = X.copy()
data['price'] = y

# Correlation with target
corr = data.corr()

corr['price'].sort_values(ascending=False)
```

```python
selected_features = corr['price'][abs(corr['price']) > 0.30].index

selected_features
```

```python
selected_features = selected_features.drop('price')

X = X[selected_features]

print(X.head())
```

```python
print("Selected Features")

for i in X.columns:
    print(i)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training Shape :", X_train.shape)
print("Testing Shape :", X_test.shape)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)

X_test_scaled = scaler.transform(X_test)
```

```python
print(X_train_scaled[:5])
```

```python
print("Data is Ready!")

print("Training Samples :", len(X_train))
print("Testing Samples :", len(X_test))
```

```python
from sklearn.linear_model import LinearRegression

from sklearn.tree import DecisionTreeRegressor

from sklearn.ensemble import RandomForestRegressor

from sklearn.ensemble import GradientBoostingRegressor

from sklearn.svm import SVR
```

```python
from sklearn.metrics import r2_score

from sklearn.metrics import mean_absolute_error

from sklearn.metrics import mean_squared_error
```

```python
results = []
```

```python
from sklearn.linear_model import LinearRegression

lr = LinearRegression()

lr.fit(X_train, y_train)

lr_pred = lr.predict(X_test)

lr_r2 = r2_score(y_test, lr_pred)
lr_mae = mean_absolute_error(y_test, lr_pred)
lr_mse = mean_squared_error(y_test, lr_pred)

results.append(["Linear Regression", lr_r2, lr_mae, lr_mse])

print("Linear Regression Completed")
```

```python
from sklearn.tree import DecisionTreeRegressor

dt = DecisionTreeRegressor(random_state=42)

dt.fit(X_train, y_train)

dt_pred = dt.predict(X_test)

dt_r2 = r2_score(y_test, dt_pred)
dt_mae = mean_absolute_error(y_test, dt_pred)
dt_mse = mean_squared_error(y_test, dt_pred)

results.append(["Decision Tree", dt_r2, dt_mae, dt_mse])

print("Decision Tree Completed")
```

```python
from sklearn.ensemble import RandomForestRegressor

rf = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)

rf.fit(X_train, y_train)

rf_pred = rf.predict(X_test)

rf_r2 = r2_score(y_test, rf_pred)
rf_mae = mean_absolute_error(y_test, rf_pred)
rf_mse = mean_squared_error(y_test, rf_pred)

results.append(["Random Forest", rf_r2, rf_mae, rf_mse])

print("Random Forest Completed")
```

```python
from sklearn.ensemble import GradientBoostingRegressor

gb = GradientBoostingRegressor(random_state=42)

gb.fit(X_train, y_train)

gb_pred = gb.predict(X_test)

gb_r2 = r2_score(y_test, gb_pred)
gb_mae = mean_absolute_error(y_test, gb_pred)
gb_mse = mean_squared_error(y_test, gb_pred)

results.append(["Gradient Boosting", gb_r2, gb_mae, gb_mse])

print("Gradient Boosting Completed")
```

```python
from sklearn.svm import SVR

svr = SVR(kernel='rbf')

svr.fit(X_train_scaled, y_train)

svr_pred = svr.predict(X_test_scaled)

svr_r2 = r2_score(y_test, svr_pred)
svr_mae = mean_absolute_error(y_test, svr_pred)
svr_mse = mean_squared_error(y_test, svr_pred)

results.append(["Support Vector Regressor", svr_r2, svr_mae, svr_mse])

print("SVR Completed")
```

```python
results_df = pd.DataFrame(
    results,
    columns=["Model", "R2 Score", "MAE", "MSE"]
)

results_df
```

```python
results_df = results_df.sort_values(
    by="R2 Score",
    ascending=False
)

results_df
```

```python
best_model = results_df.iloc[0]

print("Best Model")

print(best_model)
```

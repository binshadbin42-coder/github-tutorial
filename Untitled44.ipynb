```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error, r2_score

import tensorflow as tf
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau

print("TensorFlow Version:", tf.__version__)
```

```python
diabetes = load_diabetes()

X = diabetes.data
y = diabetes.target

print("Feature shape:", X.shape)
print("Target shape:", y.shape)

print("\nFeature Names:")
print(diabetes.feature_names)
```

```python
df = pd.DataFrame(
    X,
    columns=diabetes.feature_names
)

df["target"] = y

df.head()
```

```python
print("Dataset Information:")
df.info()

print("\nDataset Shape:")
print(df.shape)

print("\nStatistical Summary:")
display(df.describe())
```

```python
print("Missing Values:")
print(df.isnull().sum())
```

```python
df = df.fillna(df.mean())

print("Missing values after handling:")
print(df.isnull().sum())
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df["target"], bins=30)

plt.title("Distribution of Diabetes Progression")
plt.xlabel("Target")
plt.ylabel("Frequency")

plt.show()
```

```python
df.hist(
    figsize=(15, 12),
    bins=30
)

plt.suptitle("Feature Distributions")

plt.show()
```

```python
plt.figure(figsize=(12, 8))

sns.heatmap(
    df.corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")

plt.show()
```

```python
features = diabetes.feature_names

for feature in features:

    plt.figure(figsize=(6, 4))

    plt.scatter(
        df[feature],
        df["target"]
    )

    plt.xlabel(feature)
    plt.ylabel("Diabetes Progression")
    plt.title(f"{feature} vs Target")

    plt.show()
```

```python
X = df.drop("target", axis=1)
y = df["target"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

```python
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)

X_test_scaled = scaler.transform(X_test)

print("Training data after scaling:")
print(X_train_scaled[:5])
```

```python
model = Sequential([

    Dense(
        64,
        activation="relu",
        input_shape=(X_train_scaled.shape[1],)
    ),

    Dense(
        32,
        activation="relu"
    ),

    Dense(
        16,
        activation="relu"
    ),

    Dense(
        1,
        activation="linear"
    )
])

model.summary()
```

```python
model.compile(

    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),

    loss="mse",

    metrics=["mae"]
)
```

```python
early_stopping = EarlyStopping(
    monitor="val_loss",
    patience=20,
    restore_best_weights=True
)

reduce_lr = ReduceLROnPlateau(
    monitor="val_loss",
    factor=0.5,
    patience=10,
    min_lr=0.00001
)

history = model.fit(

    X_train_scaled,
    y_train,

    validation_split=0.20,

    epochs=200,

    batch_size=16,

    callbacks=[
        early_stopping,
        reduce_lr
    ],

    verbose=1
)
```

```python
plt.figure(figsize=(8, 5))

plt.plot(
    history.history["loss"],
    label="Training Loss"
)

plt.plot(
    history.history["val_loss"],
    label="Validation Loss"
)

plt.xlabel("Epoch")
plt.ylabel("MSE Loss")
plt.title("Training vs Validation Loss")

plt.legend()

plt.show()
```

```python
y_pred = model.predict(X_test_scaled)

y_pred = y_pred.flatten()

print("Predicted values:")
print(y_pred[:10])
```

```python
mse = mean_squared_error(
    y_test,
    y_pred
)

print("Mean Squared Error:", mse)
```

```python
r2 = r2_score(
    y_test,
    y_pred
)

print("R² Score:", r2)
```

```python
print("Model Performance")
print("-----------------------")
print("MSE:", mse)
print("R² Score:", r2)
```

```python
plt.figure(figsize=(8, 6))

plt.scatter(
    y_test,
    y_pred
)

plt.xlabel("Actual Values")
plt.ylabel("Predicted Values")

plt.title("Actual vs Predicted Diabetes Progression")

plt.show()
```

```python
improved_model = Sequential([

    Dense(
        128,
        activation="relu",
        input_shape=(X_train_scaled.shape[1],)
    ),

    Dropout(0.2),

    Dense(
        64,
        activation="relu"
    ),

    Dropout(0.2),

    Dense(
        32,
        activation="relu"
    ),

    Dense(
        1,
        activation="linear"
    )
])

improved_model.summary()
```

```python
improved_model.compile(

    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.0005
    ),

    loss="mse",

    metrics=["mae"]
)
```

```python
history_improved = improved_model.fit(

    X_train_scaled,
    y_train,

    validation_split=0.20,

    epochs=300,

    batch_size=16,

    callbacks=[
        EarlyStopping(
            monitor="val_loss",
            patience=30,
            restore_best_weights=True
        ),

        ReduceLROnPlateau(
            monitor="val_loss",
            factor=0.5,
            patience=10,
            min_lr=0.00001
        )
    ],

    verbose=1
)
```

```python
y_pred_improved = improved_model.predict(
    X_test_scaled
).flatten()

mse_improved = mean_squared_error(
    y_test,
    y_pred_improved
)

r2_improved = r2_score(
    y_test,
    y_pred_improved
)

print("Improved Model Performance")
print("-----------------------------")
print("MSE:", mse_improved)
print("R² Score:", r2_improved)
```

```python
comparison = pd.DataFrame({

    "Model": [
        "Original ANN",
        "Improved ANN"
    ],

    "MSE": [
        mse,
        mse_improved
    ],

    "R2 Score": [
        r2,
        r2_improved
    ]
})

comparison
```

```python
comparison.plot(
    x="Model",
    y=["MSE", "R2 Score"],
    kind="bar",
    figsize=(8, 5)
)

plt.title("Original vs Improved ANN")

plt.ylabel("Score")

plt.xticks(rotation=0)

plt.show()
```

```python
sample = X_test.iloc[0:5]

sample_scaled = scaler.transform(sample)

predictions = improved_model.predict(
    sample_scaled
).flatten()

result = pd.DataFrame({

    "Actual": y_test.iloc[0:5].values,

    "Predicted": predictions

})

result
```

```python
# ============================================================
# AI-POWERED PARAPHRASING TOOL
# Module End Assignment 6 - GenAI
# ============================================================

# ------------------------------------------------------------
# Install missing libraries
# ------------------------------------------------------------
!pip install rouge_score
!pip install sentence-transformers
!pip install language-tool-python
!pip install transformers

# Import required libraries
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
from sentence_transformers import SentenceTransformer, util
from rouge_score import rouge_scorer
import language_tool_python
import nltk


# ------------------------------------------------------------
# 1. Download required NLTK resources
# ------------------------------------------------------------

nltk.download("punkt")


# ------------------------------------------------------------
# 2. Load T5 Paraphrasing Model
# ------------------------------------------------------------

print("Loading paraphrasing model...")

model_name = "ramsrigouthamg/t5_paraphraser"

tokenizer = AutoTokenizer.from_pretrained(model_name)

model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

print("Model loaded successfully!")


# ------------------------------------------------------------
# 3. Load Semantic Similarity Model
# ------------------------------------------------------------

print("Loading semantic similarity model...")

similarity_model = SentenceTransformer(
    "all-MiniLM-L6-v2"
)

print("Similarity model loaded successfully!")


# ------------------------------------------------------------
# 4. Grammar Checker
# ------------------------------------------------------------

print("Starting grammar checker...")

grammar_tool = language_tool_python.LanguageTool("en-US")


# ------------------------------------------------------------
# 5. Paraphrasing Function
# ------------------------------------------------------------

def paraphrase_text(text):

    # Create T5 input format
    input_text = "paraphrase: " + text

    # Convert text into tokens
    inputs = tokenizer(
        input_text,
        return_tensors="pt",
        max_length=512,
        truncation=True
    )

    # Generate paraphrase
    outputs = model.generate(
        **inputs,
        max_length=512,
        num_beams=5,
        num_return_sequences=1,
        temperature=1.5,
        repetition_penalty=1.2,
        early_stopping=True
    )

    # Convert tokens back to text
    paraphrased_text = tokenizer.decode(
        outputs[0],
        skip_special_tokens=True
    )

    return paraphrased_text


# ------------------------------------------------------------
# 6. Grammar Checking Function
# ------------------------------------------------------------

def grammar_check(text):

    matches = grammar_tool.check(text)

    return matches


# ------------------------------------------------------------
# 7. Semantic Similarity Function
# ------------------------------------------------------------

def semantic_similarity(original, paraphrased):

    original_embedding = similarity_model.encode(
        original,
        convert_to_tensor=True
    )

    paraphrased_embedding = similarity_model.encode(
        paraphrased,
        convert_to_tensor=True
    )

    similarity = util.cos_sim(
        original_embedding,
        paraphrased_embedding
    )

    return float(similarity[0][0])


# ------------------------------------------------------------
# 8. ROUGE Evaluation
# ------------------------------------------------------------

def calculate_rouge(original, paraphrased):

    scorer = rouge_scorer.RougeScorer(
        ["rouge1", "rouge2", "rougeL"],
        use_stemmer=True
    )

    scores = scorer.score(
        original,
        paraphrased
    )

    return scores


# ------------------------------------------------------------
# 9. Display Grammar Results
# ------------------------------------------------------------

def display_grammar_results(text):

    matches = grammar_check(text)

    print("\n----------------------------------------")
    print("GRAMMAR & SPELLING CHECK")
    print("----------------------------------------")

    if len(matches) == 0:

        print("No grammar or spelling errors found.")

    else:

        print(
            "Number of possible errors:",
            len(matches)
        )

        for match in matches[:10]:

            print("\nError:")
            print(match.message)

            print("Suggestion:")

            if match.replacements:
                print(match.replacements[:3])
            else:
                print("No suggestion available.")


# ------------------------------------------------------------
# 10. Main Program
# ------------------------------------------------------------

print("\n========================================")
print("     AI-POWERED PARAPHRAZING TOOL")
print("========================================")

print("\nEnter your English text below.")
print("Type 'exit' to close the program.\n")


while True:

    # Take input from user
    original_text = input("Enter text: ")

    # Exit condition
    if original_text.lower() == "exit":

        print("\nProgram closed.")
        break


    # Check empty input
    if original_text.strip() == "":

        print("Please enter some text.")

        continue


    # --------------------------------------------------------
    # Generate paraphrase
    # --------------------------------------------------------

    print("\nGenerating paraphrase...")

    paraphrased_text = paraphrase_text(
        original_text
    )


    # --------------------------------------------------------
    # Display result
    # --------------------------------------------------------

    print("\n========================================")
    print("ORIGINAL TEXT")
    print("========================================")

    print(original_text)


    print("\n========================================")
    print("PARAPHRASED TEXT")
    print("========================================")

    print(paraphrased_text)


    # --------------------------------------------------------
    # Grammar checking
    # --------------------------------------------------------

    display_grammar_results(
        paraphrased_text
    )


    # --------------------------------------------------------
    # Semantic similarity
    # --------------------------------------------------------

    similarity = semantic_similarity(
        original_text,
        paraphrased_text
    )


    print("\n----------------------------------------")
    print("SEMANTIC SIMILARITY")
    print("----------------------------------------")

    print(
        "Similarity Score:",
        round(similarity, 4)
    )


    # --------------------------------------------------------
    # ROUGE evaluation
    # --------------------------------------------------------

    rouge_scores = calculate_rouge(
        original_text,
        paraphrased_text
    )


    print("\n----------------------------------------")
    print("ROUGE EVALUATION")
    print("----------------------------------------")

    print(
        "ROUGE-1:",
        round(
            rouge_scores["rouge1"].fmeasure,
            4
        )
    )

    print(
        "ROUGE-2:",
        round(
            rouge_scores["rouge2"].fmeasure,
            4
        )
    )

    print(
        "ROUGE-L:",
        round(
            rouge_scores["rougeL"].fmeasure,
            4
        )
    )


    # --------------------------------------------------------
    # Final quality interpretation
    # --------------------------------------------------------

    print("\n----------------------------------------")
    print("QUALITY CHECK")
    print("----------------------------------------")

    if similarity >= 0.80:

        print(
            "Meaning preservation: Excellent"
        )

    elif similarity >= 0.65:

        print(
            "Meaning preservation: Good"
        )

    elif similarity >= 0.50:

        print(
            "Meaning preservation: Moderate"
        )

    else:

        print(
            "Meaning preservation: Low"
        )


    print("\n========================================")
```
